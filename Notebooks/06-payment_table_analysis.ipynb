```python
import pandas as pd
```

```python
payment_data=pd.read_csv("../Data/Processed/payments_cleaned.csv")
order_data=pd.read_csv("../Data/Processed/orders_cleaned.csv")
customer_data=pd.read_csv("../Data/Processed/customers_cleaned.csv")
```

```python
# How many total payment records are there?
Total_payment_records=payment_data["order_id"].count()
print(Total_payment_records)

# The total payments records in olist payment data is 103886 duplicates value is also presente in this
```

```python
# How many unique orders have payment records?
Total_unique_payment_records=payment_data["order_id"].nunique()
print(Total_unique_payment_records)

# There are 99440 payments records that have uique orders
```

```python
payment_data["payment_installments"].value_counts().sort_index()
```

```python
# What are the different payment methods?
payment_method=payment_data["payment_type"].unique()
payment_method

# There are in total 5 different paymenet methods presents
```

```python
# What is the total and average payment value?
Total_payment_value=payment_data["payment_value"].sum()
Avg_payment_value=payment_data["payment_value"].mean()

print(Total_payment_value)
print(Avg_payment_value)

# The total payment value is 16008872.120000001 and the avg of this payment value is 154.10038041699556
```

```python
# How frequently is each payment method used?
payment_method_freq=payment_data["payment_type"].value_counts()
payment_method_freq

# Mostly credit card payment methood is used frequently and not defined payment method is used only 3 time.
```

```python
# Which payment methods generate the highest total payment value?

payment_method_value=payment_data.groupby("payment_type")[
    ["payment_value"]
    ].sum()

payment_method_value.sort_values("payment_value",ascending=False)

# Highest  total payment value is done through credit card payment method.
```

```python
# Which payment methods have the highest average payment value?

Avg_payment_method_value=payment_data.groupby("payment_type")[
    ["payment_value"]
    ].mean()

Avg_payment_method_value.sort_values("payment_value",ascending=False)

# The payment method that generate the highest avg payment value is credit card.
```

```python
# What percentage of payments are made in installments?

Total_payment_installment=payment_data["payment_installments"].sum()
Count_payment_installment=payment_data[
    (payment_data["payment_installments"]>1)
    ].shape[0]

percent_payment_instalation=Total_payment_installment/Count_payment_installment
percent_payment_instalation
# Count_payment_installment
# 285.33488631769444 percentage of payments are made in installments.
```

```python
# Which payment methods are most commonly used for installments?
payment_method_installments=payment_data[
    payment_data["payment_installments"]>1
    ].groupby("payment_type")[
    ["payment_installments"]
    ].count()

payment_method_installments#.sort_values(
#     "payment_installments",
#     ascending=False
# )

# The payments method that are used most commonly for the installments is credit card.
```

```python
# Does a higher number of installments correspond to higher order values?

installment_order_relation=payment_data[
    ["payment_installments","payment_value"]
    ].corr()

print(installment_order_relation)

# Number of installment and order values have a weak to moderate positive linear realtion as the correlation of Number of installment with order values
# is 0.330811
```

Merge different table

```python
payments_orders_table=payment_data.merge(
    order_data[["order_id","customer_id"]],
    on="order_id",
    how="left"
)

payments_orders_customer_table=payments_orders_table.merge(
    customer_data[["customer_id","customer_state","customer_city"]],
    on="customer_id",
    how="left"
)

payments_orders_customer_table
```

```python
# Which customer states generate the highest payment value?
customer_state_payment_value=payments_orders_customer_table.groupby("customer_state")[
    ["payment_value"]
    ].sum()
customer_state_payment_value.sort_values("payment_value",ascending=False).head(1)

# The highest payment value is generated through the sao paulo state.
```

```python
# What percentage of orders have multiple payment transactions?

multiple_payment_installment=payment_data[
    (payment_data["payment_installments"]>1)
    ].shape[0]

Total_records=payment_data["order_id"].count()

percent_orders_multiple_payment_transaction=multiple_payment_installment/Total_records*100
percent_orders_multiple_payment_transaction

# Total_records
# multiple_payment_installment

# out of 100 the percentage of orders have multiple payment transactions is 49.41763086460158
```

```python
# What is the average number of payment transactions per order?

Total_transaction=payment_data["order_id"].count()  # Do not use payment_installments for this question. An order with payment_installments = 6 still represents one payment record, not six payment transactions.
Total_orders=payment_data["order_id"].nunique()

avg_payment_transaction_per_orders=Total_transaction/Total_orders

avg_payment_transaction_per_orders.astype(int)

# On every order the avg number of payment transaction done is 1
```
