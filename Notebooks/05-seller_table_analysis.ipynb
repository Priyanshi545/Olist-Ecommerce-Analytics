```python
import pandas as pd
```

```python
seller_data=pd.read_csv("../Data/Processed/sellers_cleaned.csv")
order_items_data=pd.read_csv("../Data/Processed/order_items_cleaned.csv")
order_data=pd.read_csv("../Data/Processed/orders_cleaned.csv")
customer_data=pd.read_csv("../Data/Processed/customers_cleaned.csv")
```

```python
# How many unique sellers are there?
unique_sellers=seller_data["seller_id"].nunique()
unique_sellers

# Total unique sellers are 3095 in olist seller data
```

```python
# How many sellers are there in each state?

seller_state=seller_data.groupby("seller_state")[
    ["seller_id"]
    ].nunique()

seller_state.sort_values("seller_id",ascending=False)

# Sao paulo state have highest number of unique sellers 1849 while PA have only 1 unique seller in total
```

```python
# Which states have the highest number of sellers?
highest_seller_state=seller_data.groupby("seller_state")[
    ["seller_id"]
    ].nunique()

highest_seller_state.sort_values("seller_id",ascending=False).head(1)

# Sao paulo state have highest numbers of sellers among all the other states
```

```python
# What are the top 10 cities by seller count?
Top_10_seller_city=seller_data.groupby("seller_city")[
    ["seller_id"]
    ].nunique()

Top_10_seller_city.sort_values("seller_id",ascending=False).head(10)

# These are the top 10 city that have the highest seller count
```

```python
seller_order_items=seller_data.merge(
    order_items_data[["seller_id","price","freight_value","order_id","product_id"]],
    on="seller_id",
    how="left"
)
seller_order_items
```

```python
# Which sellers generate the highest revenue?
seller_revenue=seller_order_items.groupby("seller_id")[
    ["price","freight_value"]
    ].sum()
seller_revenue.sort_values("price",ascending=False).head(1)
```

```python
# Which sellers sell the highest number of items?
seller_items=seller_order_items.groupby("seller_id")[
    ["product_id"]
    ].nunique()

seller_items.sort_values("product_id",ascending=False).head(1)
```

```python
# Which sellers receive the highest number of orders?
seller_items=seller_order_items.groupby("seller_id")[
    ["order_id"]
    ].count()

seller_items.sort_values("order_id",ascending=False).head(1)
```

```python
# Which sellers have the highest average order value?

order_value=(order_items_data.groupby("seller_id")
             .agg(
                 total_order=("order_id","count"),
                 order_value=("price","sum")
             )
            )
order_value["avg_order_value"]=order_value["total_order"]/order_value["order_value"]
order_value.sort_values("avg_order_value",ascending=False).head(1)
```

```python
# Which sellers contribute the most to total marketplace revenue?

total_marketplace_revenue=seller_order_items.groupby("seller_id")[
    ["price"]
    ].sum()

total_marketplace_revenue.sort_values("price",ascending=False).head(1)
```

```python
# Which sellers have high sales volume but low revenue?

seller_sales_revenue=(seller_order_items.groupby("seller_id")
                      .agg(
                          sales_volume=("order_id","count"),
                          revenue=("price","sum")
                      )
                     )
high_sales_low_revenue=seller_sales_revenue[
    (seller_sales_revenue["sales_volume"]>seller_sales_revenue["sales_volume"].mean())&
    (seller_sales_revenue["revenue"]<seller_sales_revenue["revenue"].mean())
    ]

high_sales_low_revenue.head(1)

# sales_mean=seller_sales_revenue["sales_volume"].mean()
# revenue_mean=seller_sales_revenue["revenue"].mean()
# print(sales_mean)  #36.39741518578352
# print(revenue_mean)  # 4391.484232633279
```

```python
sales_revenue=seller_sales_revenue[
    ["sales_volume","revenue"]
    ].corr()
sales_revenue

# sales volume and revenue appear to be dependent on each other because the relationship between sales and revenue is positive relationship(0.795955)
```

```python
# Which seller states generate the highest revenue?
seller_state_revenue=seller_order_items.groupby("seller_state")[
    ["price"]
    ].sum()
seller_state_revenue.sort_values("price",ascending=False).head(1)
```

```python
columns=[  
    "order_purchase_timestamp",            
     "order_approved_at",
     "order_delivered_carrier_date",  
     "order_delivered_customer_date",    
     "order_estimated_delivery_date"
]
for col in columns:
    order_data[col]=pd.to_datetime(
        order_data[col],format="mixed"
    )
    
order_data.info()
```

```python
seller_items=seller_order_items.merge(
    order_items_data[["order_id","seller_id"]],
    on="seller_id",
    how="left"
)
seller_items_order=seller_order_items.merge(
    order_data[["order_id","delivery_delay_days","order_delivery_time"]],
    on="order_id",
    how="left"
)
seller_items_order
```

```python
# Which sellers have the highest average delivery delay?

seller_avg_delivery_delay=seller_items_order.groupby("seller_id")[
    ["delivery_delay_days"]
    ].mean()

seller_avg_delivery_delay.sort_values("delivery_delay_days", ascending=False).head(1)
```

```python
# Which sellers have the longest average delivery time?
seller_delivery_time=seller_items_order.groupby("seller_id")[
    ["order_delivery_time"]
    ].mean()
seller_delivery_time.sort_values("order_delivery_time",ascending=False).head(1)
```

```python
# Are high-revenue sellers also good at delivery performance?
seller_revenue_delivery_performance=(seller_items_order.groupby("seller_id")
                                     .agg(
                                         revenue=("price","sum"),
                                         delivery_performance=("order_delivery_time","mean")
                                     )
                                    )
high_revenue_low_delivery=seller_revenue_delivery_performance[
    (seller_revenue_delivery_performance["revenue"]>seller_revenue_delivery_performance["revenue"].mean())&
    (seller_revenue_delivery_performance["delivery_performance"]<seller_revenue_delivery_performance["delivery_performance"].mean())
    ]

print(high_revenue_low_delivery)

revenue_delivery=seller_items_order[
    ["price","order_delivery_time"]
    ].corr()

print(revenue_delivery)

seller_mean=seller_revenue_delivery_performance["revenue"].mean()
seller_delivery_mean=seller_revenue_delivery_performance["delivery_performance"].mean()
print(seller_mean)
print(seller_delivery_mean)

# Revenue and delivery performance appear to be largely independent in this dataset. While some high-revenue sellers achieve faster-than-average 
# delivery times, the overall relationship between product price and delivery time is negligible (r = 0.062), suggesting that product price is not a 
# meaningful predictor of delivery duration.
```
