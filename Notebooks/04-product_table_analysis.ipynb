```python
category_revenue=product_data.merge(
    order_items_data[["product_id","price","freight_value","order_id"]],
    on="product_id",
    how="left"
)
category_revenue
```

```python
# Which product categories generate the highest revenue?

category_highest_revenue=category_revenue.groupby("product_category_name")[
    ["price"]
    ].sum()
category_highest_revenue.sort_values("price",ascending=False)

# beleza_saude product category generate the highest revenue 1258681.34 among all the other category while seguros_e_servicos generate lowest 
# revenue 283.29
```

```python
# Which product categories have the highest number of items sold?

category_highest_items_sold=product_data.groupby("product_category_name")[
    ["product_id"]
    ].count()
category_highest_items_sold.sort_values("product_id",ascending=False)

# Highest number of items sold in cama_mesa_banho category that is 3029 while cds_dvds_musicais sold less items that is 1
```

```python
# Which products generate the highest revenue?

product_high_revenue=category_revenue.groupby("product_id")[
    ["price"]
    ].sum()
product_high_revenue.sort_values("price",ascending=False).head(1)
    
```

```python
# Which products have the highest sales volume?

product_high_sales_volume=category_revenue.groupby("product_id")[
    ["order_id"]
    ].count()
product_high_sales_volume.sort_values("order_id",ascending=False).head(1)
    
```

```python
# Which categories have a high number of products but low sales volume?

category_analysis=(category_revenue.groupby("product_category_name")
                   .agg(
                       total_products=("product_id","count"),
                       sales_volume=("order_id","count")
                   )
                  )
high_product_low_Sales=category_analysis[
    (category_analysis["total_products"]>category_analysis["total_products"].mean()) &
    (category_analysis["sales_volume"]<category_analysis["sales_volume"].mean())
    ]

high_product_low_Sales.sort_values(
    "total_products",
    ascending=False
)

# suprisingly there is no product category which have high number of products but low sales volume
```

```python
# Which categories generate high revenue relative to their number of products?

category_product=(category_revenue.groupby("product_category_name")
                  .agg(
                      revenue=("price","sum"),
                       number_of_products=("product_id","nunique")
                  
                 )
                 )
category_product["revenue_per_product"]=(
    category_product["revenue"]/category_product["number_of_products"]
)
category_product.sort_values("revenue_per_product",ascending=False)

# pcs product_category has high revenue per products while flores have lowest revenue per products
```

```python
# Does product weight affect freight value?
weight_affect_freight = category_revenue[
    ["product_weight_g", "freight_value"]
].corr()

weight_affect_freight

# # |     Correlation | Interpretation                 |
# | --------------: | ------------------------------ |
# | Close to **+1** | Strong positive relationship   |
# | Around **+0.5** | Moderate positive relationship |
# |    Around **0** | Little/no linear relationship  |
# | Around **-0.5** | Moderate negative relationship |
# | Close to **-1** | Strong negative relationship   |


# There is a positive correlation between weight and freight value of 0.757451 which indicates that the heavier the product weight is the higher the 
# shipping charge/freight values will be
```

```python
# Do products with more photos generate higher sales?

product_photos_relation_sales=category_revenue[
    ["product_photos_qty","price"]
    ].corr()

product_photos_relation_sales

# product photos has almost no linear relation with sales as there are 0.051848 correlation between them which indicates that more photos do not 
# generate the high product sales
```

```python
# Which categories have the highest freight-to-price ratio?

category_freight_price=(category_revenue.groupby("product_category_name")
                        .agg(
                            freight=("freight_value","sum"),
                            price=("price","sum")
                        )
                       )
category_freight_price["freight_price_ratio"]=category_freight_price["freight"]/category_freight_price["price"]

category_freight_price.sort_values("freight_price_ratio",ascending=False)

# casa_conforto_2 has highest freight-to-price ratio while pcs has highest ratio
```

```python
# Which product characteristics are associated with higher revenue?

product_revenue_characteristics = (category_revenue[
    [
        "product_weight_g",
        "product_length_cm",
        "product_height_cm",
        "product_width_cm",
        "product_photos_qty",
        "product_name_lenght",
        "product_description_lenght",
        "price"
    ]
].corr()
                                  )

product_revenue_characteristics["price"].sort_values(ascending=False)

# product_weight has the strongest positive correlation with revenue among all the product characteristics while product height,description,width and
# length has weak correlation and product photos and produt name length has no linear relation with revenue
```
