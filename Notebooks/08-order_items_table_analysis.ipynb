```python
import pandas as pd
import numpy as np
```

```python
order_items_data=pd.read_csv("../Data/Processed/order_items_cleaned.csv")
product_data=pd.read_csv("../Data/Processed/products_cleaned.csv")
customer_data=pd.read_csv("../Data/Processed/customers_cleaned.csv")
order_data=pd.read_csv("../Data/Processed/orders_cleaned.csv")
```

```python
# How many total order-item records are there?
Total_order_items=order_items_data["order_id"].count()
print(Total_order_items)

# There is 112650 order items records which is order by the customers among these there are repeated products also
```

```python
# What is the average product price and freight value?

Avg_product_price = order_items_data["price"].mean()
Avg_freight_value=order_items_data["freight_value"].mean()
print(Avg_product_price)
print(Avg_freight_value)

# The avg product price per order items is 120.653 and freight/shipping charge is 19.990 these avg includes the repeated items also
```

```python
# "What is the average total price per unique product?"
Total_price_per_product = (
    order_items_data
    .groupby("product_id")["price"]
    .sum()
)

Avg_product_price = Total_price_per_product.mean()

Avg_product_price

# avg of unique product price is 412.480
# "avg of unique product price" That could be misunderstood as the average unit price of unique products.

# It actually means: On average, each unique product generated 412.48 in cumulative product sales across all its order-item transactions.
```

```python
# What is the average number of items per order?
item_per_order=(order_items_data.groupby("order_id")
["order_item_id"].count()
               )
Avg_item_per_order=item_per_order.mean()
Avg_item_per_order

# on avg the number of items per order is 1.1417
```

```python
# What percentage of orders contain multiple items?

Total_orders=order_items_data["order_id"].nunique()

item_per_order=(order_items_data.groupby("order_id")
["product_id"].count()
               )

item_multiple_order=item_per_order[item_per_order>1].count()
item_multiple_order_percent=item_multiple_order/Total_orders*100

print(item_multiple_order_percent)

# 9.93% of order contain the multiple items
```

```python
# What is the total product revenue? sales kpi
Total_product_revenue=order_items_data["price"].sum()
Total_product_revenue

# Total product sales generate 13591643.7 revenue
```

```python
# What is the total revenue including freight?
Total_revenue=order_items_data["price"].sum()
Total_freight_value=order_items_data["freight_value"].sum()
Total_sales=Total_revenue+Total_freight_value
print(Total_sales)

# Total order value including freight = 15.84M
```

```python
order_items_product=order_items_data.merge(
    product_data[["product_id","product_category_name"]],
    on="product_id",
    how="left"
)

order_items_order=order_items_product.merge(
    order_data[["order_id","customer_id","order_status"]],
    on="order_id",
    how="left"
)

order_items_customer=order_items_order.merge(
    customer_data[["customer_id","customer_unique_id","customer_state","customer_city"]],
    on="customer_id",
    how="left"
)

order_items_customer
```

```python
# Which product categories generate the highest revenue?
highest_revenue_category=(order_items_product.groupby("product_category_name")
["price"].sum()
)
highest_revenue_category.sort_values(ascending=False)

# beleza_saude generate highest categories revenue as it is possible that it avg product price is highest among all the other category while 
# seguros_e_servicos generate the lowest revenue as it avg product price is lowest among all while the futher analaysis can be donento prove this 
# point and other factors is also responsible for the category revenue

# Beleza_saude generates the highest product revenue among the categories, with approximately 1.26M in sales. Further analysis of sales volume and 
# average selling price is required to determine the main drivers of its high revenue.
```

```python
# Which product categories have the highest number of items sold?
highest_item_sold_category=order_items_product.groupby("product_category_name")[
    ["product_id"]
    ].count()
highest_item_sold_category.sort_values("product_id",ascending=False)

# Cama_mesa_banho has the highest sales volume, with 11,115 product items sold.
```

```python
# Which sellers generate the highest revenue?
seller_highest_revenue=order_items_data.groupby("seller_id")[
    ["price"]
    ].sum()
seller_highest_revenue.sort_values("price",ascending=False).head(1)
```

```python
# Which products generate the highest revenue?
product_highest_revenue=order_items_data.groupby("product_id")[
    ["price"]
    ].sum()
product_highest_revenue.sort_values("price",ascending=False).head(1)
```

```python
# Which categories have high sales volume but relatively low revenue?

category_distribution = (
    order_items_product
    .groupby("product_category_name")
    .agg(
        sales_volume=("product_id", "count"),
        revenue=("price", "sum")
    )
)

category_distribution.sort_values(
    ["sales_volume", "revenue"],
    ascending=[False, True]
)                  

# while seguros_e_servicos has lowest sales but lowest revenue as sales factor is not only helful in generatinng the highest revenue for the particular 
# category while beleza_saude has highest sales but  not having highest revenue  This indicates that sales volume alone does not determine revenue; 
# the selling price of products also plays an important role.
```

```python
# Which sellers have high sales volume but low average order value?

seller_distribution=(order_items_data.groupby("seller_id")
                     .agg(
                         sales_volume=("product_id","count"),
                         total_orders=("order_id","nunique"),
                         total_revenue=("price","sum")
                     )
                    )

seller_distribution["Avg_order_value"]=(seller_distribution["total_revenue"]/seller_distribution["total_orders"])

high_volume_low_aov = seller_distribution[
    (seller_distribution["sales_volume"] > seller_distribution["sales_volume"].mean()) &
    (seller_distribution["Avg_order_value"] < seller_distribution["Avg_order_value"].mean())
]
high_volume_low_aov.sort_values(
    "sales_volume",
    ascending=False
).head(1)

# These sellers have high sales volume but relatively low average order value. This suggests that they sell a large number of items, but the average 
# monetary value generated per order is comparatively low.
```

```python
# Which customer states generate the highest order-item revenue?
customer_state_revenue=order_items_customer.groupby("customer_state")[
    ["price"]
    ].sum()

customer_state_revenue.sort_values("price",ascending=False)

# sao paulo generate the highest revenue
```

```python
customer_state_revenue_analysis=(order_items_customer.groupby(["customer_state","product_category_name"])
    .agg(
        revenue=("price","sum"),
        total_product=("product_id","count")
    )
                       )
customer_state_revenue_analysis.sort_values(
    ["revenue","total_product"],
    ascending=[False,False]
)

# Sao paulo genertate highest state revenue because of its cama_mesa_banho product category as its revenue is highest among all other state and this 
# state sold highest products quantity

# São Paulo generates the highest overall order-item revenue. Within São Paulo, cama_mesa_banho is the highest-revenue product category, generating 
# 478,284.52. This category is therefore an important contributor to revenue in São Paulo, although the state's overall revenue is generated across 
# multiple categories.
```

```python
# Which categories have the highest average freight value?
category_avg_freight_value=order_items_product.groupby("product_category_name")[
    ["freight_value"]
    ].mean()
category_avg_freight_value.sort_values("freight_value",ascending=False)

# pcs has the highest avg freight value
```

```python
# Which category contributes most to each customer state's revenue?

state_category_revenue = (
    order_items_customer
    .groupby(["customer_state", "product_category_name"])["price"]
    .sum()
    .reset_index()
)

top_category_by_state = (
    state_category_revenue
    .loc[
        state_category_revenue
        .groupby("customer_state")["price"]
        .idxmax()   # idxmax() means: Find the index/row position where the maximum value occurs.
    ]
    .sort_values("price", ascending=False)
)

top_category_by_state


# Remember, idxmax() gave us the row indexes where each state's revenue is highest..loc[] says:Go back to state_category_revenue and select those specific rows.
```
