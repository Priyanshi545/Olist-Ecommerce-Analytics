```python
customer_with_orders=customer_data.merge(
    order_data[["customer_id","order_id","order_status"]],
    on="customer_id",
    how="left"
)
customer_with_orders
```

```python
# How many unique customers made only one purchase?

order_per_customer=customer_with_orders.groupby("customer_unique_id")[
["order_id"]
    ].nunique()
print(order_per_customer.value_counts())

One_purchase_customer=order_per_customer[order_per_customer==1].count().unique()
print(One_purchase_customer)

# repeat_customer=order_per_customer[order_per_customer>1].sum().unique()
# print(repeat_customer) #6342

# Most of the orders purchased by customer only at one time so to increased the sell we have to change the marketing strategies in such a way that the
# customers start purchasing products or placed orders more then one time
```

```python
# How many customers are repeat customers?

repeat_customer=(order_per_customer>1).sum()
print(repeat_customer)
```

```python
# What is the repeat-customer rate?
repeat_customer_rate=repeat_customer/Total_customers*100
repeat_customer_rate

# out of 100 there is only 3% repeat customers rate so to increased the sales and profit our future goal can be to increased the customer retention and also 
# considered other factors also along with this
```

```python
# What is the average number of orders per customer?
Avg_customer_orders=customer_with_orders.groupby("customer_unique_id")[
["order_id"]
    ].nunique()
Avg_customer_orders.mean()
```

```python
customer_order_revenue=customer_with_orders.merge(
    order_items_data[["order_id","price","freight_value"]],
    on="order_id",
    how="left"
)
customer_order_revenue
```

```python
# What is the revenue generated by customers in each state?
State_customer_revenue=(
    customer_order_revenue.groupby("customer_state")
.agg(
customer_count=("customer_unique_id","nunique"),
Revenue=("price","sum")
)
)
State_customer_revenue.sort_values(["customer_count","Revenue"],ascending=False)

# SP state contribute to generate the highest revenue because its has high customer base and rr generate the lowest revenue as it customer base is low
# so rr has to change its marketing,logistic startegies and build a new strategies that will attract more customers such as giving some discounts ,
# offers ,free gifts or gift vouchers ets as customer count along is not responsible for the revenue generation other factors also play an inportant 
# role such as Revenue can depend on number of customers,purchase frequency,average order value,product mix etc
```

```python
# Which state has the highest average revenue per customer?

Revenue_per_customer=(
    customer_order_revenue
        .groupby(["customer_unique_id","customer_state"])
["price"].sum().reset_index()
)
# print(Revenue_per_customer)

state_avg_revenue=(
    Revenue_per_customer
    .groupby("customer_state")
        ["price"].mean().sort_values(ascending=False)
)
# print(state_avg_revenue)
customer_order_revenue
# This tell us something interesting as pb state although generate less revenue by its total customer base but its avg revenue per customer is 
# high as each customer spend more money on individual products.This indicates that customers in Paraíba spend more on average than customers in other
# states. This makes PB an interesting market for customer-value analysis.

# reset index converts customer_unique_id and customer_state back into columns, and price remains a column.
```

```python
# Which customers generated the highest revenue?
Highest_revenue_customer=customer_order_revenue.groupby("customer_unique_id")[
    ["price"]
    ].sum()
Highest_revenue_customer.sort_values("price",ascending=False).head(1)
```

```python
# Which state has the highest number of repeat customers?
order_per_customer = (
    customer_with_orders
        .groupby(["customer_unique_id", "customer_state"])["order_id"]
        .count()
        .reset_index()
)

repeat_customer = order_per_customer[
    order_per_customer["order_id"] > 1
]

# Repeat customers by state
state_repeat_customer = (
    repeat_customer
        .groupby("customer_state")["customer_unique_id"]
        .nunique()
)

# Total customers by state
total_customer_state = (
    customer_with_orders
        .groupby("customer_state")["customer_unique_id"]
        .nunique()
)

# One table
state_repeat_customer = (
    pd.DataFrame({
        "repeat_customer": state_repeat_customer,
        "repeat_customer_rate": (
            state_repeat_customer / total_customer_state * 100
        )
    })
    .sort_values("repeat_customer", ascending=False)
)

state_repeat_customer.sort_values(["repeat_customer","repeat_customer_rate"],ascending=False)

## sp state has higest no of repeat customers which tells that its marketing and logistic strategy is good as they understand the customer need very 
# well while rr has lowest repeat customers as they have to work on their strategies and try to understand the customer needs.
# Problem:- You wrote that this proves:SP's marketing and logistics strategy is good.It does not prove that There could be many reasons:
# larger population,greater ecommerce adoption,urbanization,income differences,product availability,customer acquisition,geography

# Better conclusion:-São Paulo has the largest absolute number of repeat customers, reflecting its much larger customer base. To assess loyalty fairly
# across states, repeat-customer rate should be compared rather than absolute repeat-customer count.
```

```python
# Compare customer count, order count, and revenue by state.
customer_comparison=(customer_order_revenue.groupby("customer_state")
    .agg(
    customer_count=("customer_unique_id","nunique"),
    order_count=("order_id","nunique"),
    revenue=("price","sum")
)
                    )
customer_comparison.sort_values(["customer_count","order_count","revenue"],ascending=False)
# it clearly tell that sp state has more numbers of customers that is why the order placed is also more and the sells is also high,while rr has lower
# customer base low order placed and lowest sells.However, this table alone does not prove that the higher customer count causes the higher order 
# volume and revenue.This shows a strong association between customer base size, order volume, and total revenue across states.
```

```python
# What is the average delivery time by customer state?

order_customer=order_data.merge(
    customer_data[["customer_id","customer_unique_id","customer_state"]],
    on="customer_id",
    how="left"
)
date_columns = [
    "order_purchase_timestamp",
    "order_delivered_customer_date"
]

order_customer[date_columns] = (
    order_customer[date_columns]
    .apply(pd.to_datetime, format="mixed")
)
order_customer["delivery_time"]=(order_customer["order_delivered_customer_date"]-order_customer["order_purchase_timestamp"]).dt.days
avg_delivery_time=order_customer.groupby("customer_state")[
["delivery_time"]
    ].mean()
avg_delivery_time.sort_values("delivery_time",ascending=False)

# RR state take more number of days to delivered the orders to the customers while sp state delivered order faster.The relationship between delivery 
# speed and customer volume/revenue would require further analysis before claiming that faster delivery causes higher customer retention or revenue.

# "SP has both a larger customer base and faster delivery times; further analysis is required to determine whether delivery performance contributes 
# to customer retention."
```
