```python
import pandas as pd
import numpy as np
```

```python
review_data=pd.read_csv("../Data/Processed/reviews_cleaned.csv")
order_data=pd.read_csv("../Data/Processed/orders_cleaned.csv")
customer_data=pd.read_csv("../Data/Processed/customers_cleaned.csv")
order_items_data=pd.read_csv("../Data/Processed/order_items_cleaned.csv")
product_data=pd.read_csv("../Data/Processed/products_cleaned.csv")
```

```python
# How many total reviews are there?

Total_reviews=review_data["review_id"].count()
Total_reviews

# There are 99,224 review records in the dataset.
```

```python
missing_values=review_data[
    (review_data["review_comment_title"].isna())|
    (review_data["review_comment_message"].isna())
    ]
missing_values
```

```python
# What is the average review score?
avg_review_score=review_data["review_score"].mean()
avg_review_score

# The avg review score is 4.08642062404257
```

```python
# What is the distribution of review scores?

review_distribution=review_data["review_score"].value_counts().sort_index()
review_distribution

# The review score is distributed on the range of 1 to 5 as 5 star review has the highest score that is  57328 and 3 start review has lowest score that
# is 3151
```

```python
conditions=[review_data["review_score"]<=3,review_data["review_score"]>3]
choices=["Negative","Positive"]
review_data["review_type"]=np.select(conditions,choices,default="Unknown")
review_data
```

```python
# What percentage of reviews are positive vs negative?

Total_reviews=review_data["review_score"].count()

Positive_review=review_data[
    (review_data["review_type"]=="Positive")
    ].shape[0]

Negative_review=review_data[
    (review_data["review_type"]=="Negative")
    ].shape[0]

Positive_review_percent=Positive_review/Total_reviews*100
Negative_review_percent=Negative_review/Total_reviews*100

print(Positive_review)
print(Negative_review)
print(Total_reviews)
print(Positive_review_percent)
print(Negative_review_percent)

# Based on the definition used in this analysis, reviews with scores of 4 or 5 are classified as positive, while scores of 1 to 3 are classified as 
# negative. Positive reviews represent a larger proportion of customer feedback, indicating generally positive customer satisfaction.
```

```python
# What percentage of reviews contain written comments?
Total_reviews_comments=review_data["review_comment_message"].count()

review_contain_comment=review_data.shape[0]
review_contain_comment_percent=Total_reviews_comments/review_contain_comment*100
print(review_contain_comment_percent)
print(Total_reviews)

# There are 41.29746835443038% of reviews in which customers write the written comments.This means less than half of customers provide detailed 
# textual feedback, so review scores are more widely available than written comments for analyzing customer satisfaction.
```

Merge different tables

```python
review_orders=review_data.merge(
    order_data[["order_id","customer_id","order_estimated_delivery_date","order_delivered_customer_date","delivery_delay_days","delivery_status"]],
    on="order_id",
    how="left"
)

review_customers=review_orders.merge(
    customer_data[["customer_id","customer_state","customer_city"]],
    on="customer_id",
    how="left"
)

review_order_items_data=review_customers.merge(
order_items_data[["order_id","product_id","seller_id","price","freight_value"]],
                      on="order_id",
                      how="left"
                      )

    
review_products_data=review_order_items_data.merge(
    product_data[["product_id","product_category_name"]],
    on="product_id",
    how="left"
)

review_products_data
```

```python
# Does late delivery lead to lower review scores?

delivery_review=review_orders[
    ["delivery_delay_days","review_score"]
    ].corr()
delivery_review

# The correlation between delivery delay and review score is approximately -0.267, indicating a weak negative relationship. As delivery delays 
# increase, review scores tend to decrease. However, the relationship is not strong, which suggests that other factors besides delivery performance 
# also influence customer satisfaction.
```

```python
# What is the average review score for early, on-time, and late deliveries?

avg_early_score=review_orders[
    review_orders["delivery_status"]=="early"
    ]["review_score"].mean()

avg_late_score=review_orders[
    review_orders["delivery_status"]=="late"
    ]["review_score"].mean()

avg_on_time_score=review_orders[
    review_orders["delivery_status"]=="on-time"
    ]["review_score"].mean()

print(avg_early_score)
print(avg_late_score)
print(avg_on_time_score)

# Compare the average review scores across early, on-time, and late deliveries. If late deliveries have the lowest average review score, this provides
# further evidence that delivery performance negatively affects customer satisfaction. Early and on-time deliveries are expected to have relatively 
# higher satisfaction scores.
```

```python
# Which customer states have the lowest average review scores?

state_review_score=review_customers.groupby("customer_state")[
    ["review_score"]
    ].mean()

state_review_score.sort_values("review_score")

# RR state generate the lowest avg review score that is 3.608696 while AP generate the highest 4.194030. States with lower review scores may require 
# further investigation into factors such as delivery performance, seller quality, product quality, or customer experience.
```

```python
# Which customer states have the highest percentage of negative reviews?

state_negative_review = review_customers[
    review_customers["review_type"] == "Negative"
].groupby("customer_state").size()

Total_state_review=review_customers.groupby(
    "customer_state"
).size()

state_negative_review_percent=state_negative_review/Total_state_review*100

state_negative_review_percent

# RR state have 39.130435% of negative review which is highest among all the state while AM have the lowest negative review. This suggests that 
# customer satisfaction may be weaker in this state. However, the total number of reviews should also be considered because a state with a small 
# number of reviews can show a high negative-review percentage.
```

```python
# Which product categories have the highest average review scores?

product_category_review_score=(review_products_data.groupby("product_category_name")
    .agg(
        Score=("review_score","mean"),
        Total_number_score=("review_score","count")
    )
                              )
                        
product_category_review_score.sort_values("Score",ascending=False)

# The product category cds_dvds_musicais have the highest avg review score 4.642857. This suggests strong customer satisfaction for this category. 
# However, the number of reviews should also be considered because a high average score based on a small number of reviews may not represent overall
# category performance.
```

```python
# Which sellers have the highest average review scores?

seller_review_score=review_products_data.groupby("seller_id")[
    ["review_score"]
    ].mean()

seller_review_score.sort_values("review_score",ascending=False)

# There are multiple sellers that have the highest avg review scores of 5. However, average review score alone may be misleading because some sellers
# may have received very few reviews. Therefore, seller review volume should be considered before concluding which sellers consistently deliver the 
# best customer experience.
```

```python
# Which sellers have both high revenue and low review scores?

seller_revenue_review_score=(review_order_items_data.groupby("seller_id")
                             .agg(
                                 Revenue=("price","sum"),
                                 Review_score=("review_score","mean")
                             )
                            )
seller_high_revenue_low_review_score=seller_revenue_review_score[
    (seller_revenue_review_score["Revenue"]>seller_revenue_review_score["Revenue"].mean())&
    (seller_revenue_review_score["Review_score"]<seller_revenue_review_score["Review_score"].mean())
    ]

seller_high_revenue_low_review_score
    
# These sellers generate above-average revenue but receive below-average review scores. They are an important business segment because strong sales do
# not necessarily indicate strong customer satisfaction. The business should investigate whether delivery issues, product quality, seller service, or 
# other factors are contributing to their lower ratings.
```

```python
# Do higher-priced products receive higher or lower review scores?

# review_type_customers.drop("review_type_y",inplace=True,axis=1)
order_value_review_score=review_products_data[
    ["price","review_score"]
    ].corr()

order_value_review_score


# The correlation between product price and review score is -0.003941, indicating virtually no linear relationship between the two variables. Higher-
# priced products do not appear to receive systematically higher or lower review scores.Product price does not appear to have a meaningful association
# with review scores.
```

```python
# Do higher-value orders receive higher or lower review scores?

order_value=order_items_data.groupby("order_id").agg(
    Order_value=("price","sum")
)

order_value_review=review_data.merge(
    order_value,
    on="order_id",
    how="left"
)

order_value_review_score=order_value_review[
    ["Order_value","review_score"]
].corr()

order_value_review_score

# The correlation between order value and review score is very close to zero, indicating virtually no linear relationship. Higher-value orders do not
# systematically receive higher or lower review scores. Therefore, order value alone does not appear to have a meaningful association with customer 
# satisfaction.
```

```python
# Which states have both high delivery delays and low review scores?

state_delivery_delay_review_score=(
     review_customers.groupby("customer_state")
     .agg(
         Delivery_delay=("delivery_delay_days","mean"),
         Review_score=("review_score","mean")
     )
 )
# review_orders
high_delivery_delay_low_score=(
     state_delivery_delay_review_score[
         (
             state_delivery_delay_review_score["Delivery_delay"]
             >
             state_delivery_delay_review_score["Delivery_delay"].mean()
         )
         &
         (
             state_delivery_delay_review_score["Review_score"]
             <
             state_delivery_delay_review_score["Review_score"].mean()
         )
     ]
 )

high_delivery_delay_low_score

    
```
